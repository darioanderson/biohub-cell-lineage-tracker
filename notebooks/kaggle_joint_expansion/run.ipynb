```python
import base64, io, zipfile, subprocess, sys, os
from pathlib import Path
packages = Path('/kaggle/input/models/jirkaborovec/biohubcelltrack-trackastra-artifact-and-packages/pytorch/default/1/packages/packages')
model_dir = Path('/kaggle/input/models/jirkaborovec/biohubcelltrack-trackastra-artifact-and-packages/pytorch/default/1/models/ctc')
assert packages.is_dir(), packages
assert model_dir.is_dir(), model_dir
npv = subprocess.check_output([sys.executable, '-c', 'import numpy; print(numpy.__version__)'], text=True).strip()
spv = subprocess.check_output([sys.executable, '-c', 'import scipy; print(scipy.__version__)'], text=True).strip()
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '--no-index', '--find-links', str(packages), '--no-deps', 'trackastra', 'geff', 'edt', 'lz4', 'imagecodecs', 'chardet', 'dask', 'zarr', 'numcodecs', 'donfig', 'google-crc32c', 'geff-spec', 'pydantic', 'typer', 'networkx'], check=True)
payload = ''
project = Path('/kaggle/working/trackastra_project')
project.mkdir(parents=True, exist_ok=True)
with zipfile.ZipFile(io.BytesIO(base64.b64decode(payload))) as archive:
    archive.extractall(project)
train_dir = Path('/kaggle/input/competitions/biohub-cell-tracking-during-development/train')
env = dict(os.environ)
env['PYTHONPATH'] = str(project/'src') + os.pathsep + str(project)
command = [sys.executable,str(project/'scripts/evaluate_joint_expansion.py'),str(train_dir),str(model_dir),'--manifest',str(project/'frozen_manifest.json'),'--output-dir','/kaggle/working/joint_expansion']
print('FROZEN JOINT EXPANSION START', command, flush=True)
subprocess.run(command, check=True, env=env)
print('FROZEN JOINT EXPANSION COMPLETED', flush=True)
```
