# Real detected-data parent-linker pilot
Private runtime pilot: one sample from each embryo, current DoG detector, current rich Hungarian tracker versus a fold-trained image linker. Unmatched detections remain unknown under sparse labels.


```python
%pip -q install zarr scipy scikit-image pandas
import base64, io, zipfile, subprocess, sys
from pathlib import Path
payload = ""
project = Path('/kaggle/working/detected_linker_project')
project.mkdir(parents=True, exist_ok=True)
with zipfile.ZipFile(io.BytesIO(base64.b64decode(payload))) as archive:
    archive.extractall(project)
train_dir = Path('/kaggle/input/competitions/biohub-cell-tracking-during-development/train')
assert train_dir.is_dir(), train_dir
env = dict(__import__('os').environ)
env['PYTHONPATH'] = str(project/'src')
command = [sys.executable, str(project/'scripts/evaluate_detected_linker.py'), str(train_dir), '--output-dir', '/kaggle/working/detected_linker_pilot_v2', '--cache-dir', '/kaggle/temp/detected_linker_cache', '--samples-per-embryo', '3', '--epochs', '3']
print('START', command, flush=True)
subprocess.run(command, check=True, env=env)
print('PILOT COMPLETED', flush=True)
```
